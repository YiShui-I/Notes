# Agent Skills 入门笔记
## Agent Skills 是什么
在当前的大语言模型（LLM）和 Agent 架构生态中，Agent Skills 已经从一个泛指概念（即智能体的“技能”或“工具”）演变为一个具体的开放标准（Open Standard）。

该标准最初由 Anthropic 于 2025 年末提出，现已成为 AI 领域广泛采用的规范。简单来说，Agent Skill 是一个用于封装可复用 AI 能力、专业知识和工作流的轻量级文件目录结构。 如果将 Agent 比作操作系统，那么 Agent Skills 就是可以按需挂载的“应用程序”或“知识包”。

Agent Skills本质是大模型可调用的说明文档，每个Skill定义了特定任务的执行规范，例如会议总结需要包含参会人员、议题、决定三项内容。

## 为什么需要Agent Skills
在传统的 Agent 开发中，开发者通常通过冗长的 System Prompt 或硬编码的函数调用（Function Calling）来赋予模型特定能力。但这带来了几个痛点：Context 窗口臃肿、指令难以进行版本控制、多 Agent 之间难以共享能力。

Agent Skills 通过以下方式解决了这些问题：
- 知识与逻辑的文件化： 所有的工作流和 Prompt 都可以作为标准文件存在，可以通过 Git 进行版本控制，而非硬编码在代码字符串中。
- 按需加载（Progressive Disclosure）： 文件系统级别的动态读取，极大节省了 Token 开销和 Context Window 占用。
- 跨平台与多智能体互操作： 标准化的格式让不同模型底座（如 Claude, GPT, Gemini）驱动的 Agent 能够互相识别和调用对方的能力。
